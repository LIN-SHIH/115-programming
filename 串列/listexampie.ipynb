# 簡介
串烈士一個可以儲存多個元素（整數，浮點數，或字串）的容器。
 串列中每個元素都是使用逗號隔開，左右有中括號[]。

```python
# scores 是一個串列，裡面每個元素都是整數。
scores = [78,29,56,80,99]
print(scores)
```

## 索引值
我們可以使用索引值(index)來存取串列中的元素。
注意:索引值由0開始。

```python
print(scores[0]) # 列出第1個元素
print(scores[1]) # 列出第2個元素
```

## 串列內容的個數
len(a)取得a這個串列中元素的個數。
len 是(legth)長度的意思。

```python
print(len(scores))
print(scores[4]) # 取最後一個元素
print(scores[len(scores)-1])
```

## 新增元素
使用 `append()` 新增元素到串列的尾端

```python
print(scores)
scores.append(78)
scores.append(93)
print(scores)
```

```python
print(scores)
scores[2] = 70 # 將 70 指定索引值為 2 的元素
print(scores)
```

```python
weights = [12]*6
print(weights)
```
