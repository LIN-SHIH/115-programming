# 九九乘法表
1*1=1,1*2=2,1*3=3,...
2*1=2,2*2=4,2*3=6,...
3*1=3,3*2=6,3*3=9,...

```python
for j in range(1,6):
    print(f'1*{j}={1*j}',end=',')
print()
for j in range(1,6):
    print(f'2*{j}={2*j}',end=',')
print()
for j in range(1,6):
    print(f'3*{j}={3*j}',end=',')
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```
