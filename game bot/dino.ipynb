使用pautogui完成小恐龍 在chrome中使用'chrome://dino'啟動 不斷監視畫面 讀取某位置的顏色 出現(83,83,83)的顏色 跳起來

```python
import pyautogui as pyg
pyg.sleep(2)
while True:
    im = pyg.screenshot()
    obs=False
    for i in range(15):
        c = im.getpixel((131+i,329))
        if c == (83,83,83):
            obs=True
            break
        if obs==True:
            pyg.press('space')
```
