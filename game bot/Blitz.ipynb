```python
def get_gem_position(r, c):
    """回傳第 r 列、第 c 欄寶石的 (x, y) 座標；r、c 從 0 開始。"""
    x = basex + c * gap
    y = basey + r * gap
    return x, y
```

```python
def get_gem_color(r, c, im):
    """回傳截圖 im 中第 r 列、第 c 欄寶石的顏色。"""
    x, y = get_gem_position(r, c)
    return im.getpixel((x, y))
```

```python
def can_move_left(r, c, im):
    """判斷第 r 列、第 c 欄寶石向左交換後是否會形成三連線。"""
    if not (0 <= r < 8 and 0 < c < 8):
        return False

    left_c = c - 1
    board = [[get_gem_color(row, col, im) for col in range(8)] for row in range(8)]
    board[r][c], board[r][left_c] = board[r][left_c], board[r][c]

    def forms_match(row, col):
        color = board[row][col]
        for dr, dc in ((0, 1), (1, 0)):
            count = 1
            for direction in (-1, 1):
                nr, nc = row + dr * direction, col + dc * direction
                while (0 <= nr < 8 and 0 <= nc < 8
                       and board[nr][nc] == color):
                    count += 1
                    nr += dr * direction
                    nc += dc * direction
            if count >= 3:
                return True
        return False

    return forms_match(r, c) or forms_match(r, left_c)
```

```python
import pyautogui as pyg

im = pyg.screenshot()
for row in range(8):
    for col in range(8):
        color = get_gem_color(row, col, im)
        print(f"第 {row + 1} 列、第 {col + 1} 欄：{color}")
```
